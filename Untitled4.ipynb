```python

# Import library Pandas
import pandas as pd
import numpy as np

# Membuat dataset penjualan kantin
data = {
    'Tanggal': [
        '2026-10-01', '2026-10-01', '2026-10-02',
        '2026-10-02', '2026-10-03', '2026-10-03',
        '2026-10-04', '2026-10-04', '2026-10-05',
        '2026-10-05'
    ],
    'Produk': [
        'Nasi Goreng', 'Es Teh', 'Mie Goreng',
        'Es Teh', 'Nasi Goreng', 'Roti',
        'Mie Goreng', 'Air Mineral', 'Roti',
        'Nasi Goreng'
    ],
    'Jumlah': [5, 10, 4, 8, 6, 7, 3, 12, 5, 4],
    'Harga': [
        12000, 3000, 10000, 3000, 12000,
        5000, 10000, 4000, 5000, 12000
    ]
}

df = pd.DataFrame(data)

# Menambahkan missing value dan duplikat untuk latihan
df.loc[2, 'Jumlah'] = np.nan
df.loc[6, 'Harga'] = np.nan
df = pd.concat([df, df.iloc[[0]]], ignore_index=True)

# Menyimpan dataset awal
df.to_csv('data_penjualan_kantin.csv', index=False)

print("Dataset berhasil dibuat!")
```

```python
# Menampilkan 5 data pertama
print("5 Data Pertama:")
display(df.head())

# Melihat informasi dataset
print("Informasi Dataset:")
df.info()

# Statistik deskriptif
print("Statistik Deskriptif:")
display(df.describe())
```

```python

# Memeriksa missing value
print("Missing Value:")
print(df.isnull().sum())

# Memeriksa data duplikat
print("Jumlah Duplikat:")
print(df.duplicated().sum())

# Menghapus data duplikat
df = df.drop_duplicates()

# Mengubah tipe data tanggal
df['Tanggal'] = pd.to_datetime(df['Tanggal'])

# Mengisi missing value pada Jumlah dengan median
df['Jumlah'] = df['Jumlah'].fillna(df['Jumlah'].median())

# Mengisi missing value pada Harga dengan median
df['Harga'] = df['Harga'].fillna(df['Harga'].median())

# Memastikan tipe data benar
df['Jumlah'] = df['Jumlah'].astype(int)
df['Harga'] = df['Harga'].astype(int)

# Memeriksa kembali data setelah dibersihkan
print("Missing Value Setelah Cleaning:")
print(df.isnull().sum())

print("Data Setelah Dibersihkan:")
display(df)
```

```python

# Membuat kolom turunan Total
df['Total'] = df['Jumlah'] * df['Harga']

# FILTERING
print("Transaksi di atas Rp20.000:")
filter_data = df[df['Total'] > 20000]
display(filter_data)

# SORTING
print("Urutan Total Penjualan Terbesar:")
sort_data = df.sort_values(
    by='Total', ascending=False
)
display(sort_data)

# GROUPBY DAN AGREGASI
print("Total Pendapatan Berdasarkan Produk:")
group_data = df.groupby('Produk')['Total'].sum()
display(group_data)

# Menghitung total pendapatan seluruh kantin
total_pendapatan = df['Total'].sum()
print("Total Pendapatan Kantin: Rp", total_pendapatan)
```

```python

# Menyimpan data yang sudah dibersihkan
df.to_csv('dataset_bersih.csv', index=False)

print("Dataset bersih berhasil disimpan!")
```
